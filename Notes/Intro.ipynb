### Advantages of DBMS
* Reduce Data Redundancy : A DBMS minimze duplicate data by storing information in a centralized database. 
* Maintain Data Consistency : When data is updated in 1 place , changes are reflected everywhere, which ensures consistency. 
* Data Security : DBMS allows authentication, user roles and access permissions to protect sensitive data. 
* Fast Data Retrieval : Data can be searched and retrieved quickly using Sql queries and indexing. 
* Supports multiuser access : Multiple user can access and work on the same data simultaneously without conflicts. 
* Ensures data integrity : Constraints such as primary key, foreign key, unique key,check constraint helps to maintain accurate and valid data. 
* backup and recovery : DBMS provides mechanism to backup data and recover it in case of system failure. 
* Data Sharing : Authorised users and applications can easily share and access the same data.
* Data Independence : Changes in the database structure have minimal impact on application programs.
* Better Data management : All data stored and managed in a centralised and organised manner. 
* Improve performance : Technique such as indexing and query optimization improve the performance of data retrieval. 
* Scalability : DBMS can efficiently handle growing volumes of data and increasing numbers of users. 
* Reduce application development time 

### Types of Databases: 
Databases can be classified into following categories :
1. Heirarchical Database : A heirarchical database stores data in a tree-like structure, where each parent record can have multiple child records. It follows One-to-Many Relationship. Ex : IBM-IMS. Applications : windows registry, file system directory structuree, govt administrative systems. 
2. Network Database : A network database stores data in a graph-like structure , where a record can have multiple parent and child records. It supports Many-to-Many Relationship. Ex : IDS(Integrated DataStore). Applications : Airline reservation systems,telecommunication networks, supply chain management.
3. Object Oriented Database : An OOP Db stores data in the form of objects,similar to OOP languages, each object contains data and methods. Ex : ObjectDB,DB4o . Applications: CAD/CAM Softwares, Engineering Simulations, Multimedia applications.
4. No SQL Database : A NOSQL db is designed to store semistructured and unstructured data, unlike relational db it doesn't require a fix table structure. Ex: Mongo Db, Redis, Apache cassandra , Neo4j. Applications: Social media apps, Real time chat Apps, Big data analytics, Iot , recommendation systems. 
5. Distributed Db : A distributed db stores data across multiple physical servers located at different locations while appear as a single logical db to users. Ex : Google-Spanner, Applications: global banking systems, international ecom platforms.
6. Cloud Db : Its a db hosted on cloud infrastructure and accessed over the internet, It provides automatic backup and high availability. Ex : Azure Sql Db , Google cloud Sql. Applications : Online learning platforms, SAAS.
7. Relational Db


## RDBMS 
Its a type of dbms that stores data in the form of tables(relations), consisting of rows+columns, It establish relationship b/w tables using Primary and Foreign keys to ensure data accuracy and consistency and integrity.
### Features 
1. Stores data in tables, rows and columns.
2. Support relationship b/w tables.
3. It reduce data redundancy.
4. Maintains data consistency.
5. Enforce data integrity using constraints.
6. It supports ACID transactions.
7. It supports Multiple users simultaneously.
8. Offers backup and recovery.  
Example : PostgreSQl,MySQL,Oracledb,Microsoft Sql server,IBMdb2, Maria db. 
### Applications
Banking systems , Hospital , Erp etc.



### Difference b/w DBMS and RDBMS
* Dbms is a general software used to manage db but Rdbms is an advanced type of dbms that stores data in related tables. 
* Every rdbms is a dbms, but every dbms is not rdbms. 
* In case of dbms, data may be stored in files, but in rdbms data in stored in tables.

### MySql
MySql is an open source relational database management system that uses Sql to store, manage, retrieve and manipulate data efficiently.It stores data in the form of tables consisting of rows and columns and establish relationship b/w tables using primary keys and foreign keys.  
The name MySql is consist of two parts : My - named after My widenius the daughter of Mysql cofounder **Michael Widenius**. Sql - stands for standard query language.

### Sql Vs MySql
* Sql is a language whereas Mysql is a software.
* Sql can't store data by itself whereas Mysql stores data in database and tables .
* Sql provides command such as Insert, Update, Delete whereas MySql executes Sql command and returns result. 
* Sql is a standard language supproted by many Dbms whereas MySql is a specific dbms that implements Sql.
* Sql doesn't need installation as its a language but Mysql must be installed on a server/computer before use.
