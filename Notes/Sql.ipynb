### Types of Sql languages
Sql consist of different categories of command, each category is designed to perform specific types of operations of a db, such as creating database, manipulating db, retreiving information , managing user permissions and controlling transactions.  
1. DDL(Data definition language) 
2. DML(Data Manipulation language)
3. DQl(Data Query language)
4. DCL(Data Control language)
5. TCL(Transaction Control language)


# DDL 
It is used to define,create,modify and delete the structure of database object. DDL commands deals with structure of objects.  
Database Objects are : 
* Database
* Table
* View
* Index
* Stored procedure
* Trigger 
* Functions

DDL commands interact directly with the database, It enforce an implicit commit before and after the statement. In DDL commands we cannot undo(rollback) the changes. 
Example of DDL:
1. create
2. alter
3. drop
4. truncate 
5. rename 


# DML
It deals with data only. Dml commands interact with the buffer first and then with the database. We can undo or rollback the changes.Dml commands are slower then DDL. DML commands are used to manipulate the data stored inside the database tables. It doesn't create/modify the table structure.  
Examples of DML:
1. Insert 
2. Update
3. Delete

# DQL
Data Query Language is a subset of SQL that is used to retreive data from one or more database tables. It allow users to view,search,filter data. 
Example:
1. Select 

# TCL
Transaction Control Language is a subset of SQL, that is used to manage transactions in a database. A Transaction is a sequence of one or more SQL statements that are treated as a single unit .  
A set of DML operations with commit or rollback is called Transaction.  
Every Transaction will start with DML operations and commit or rollback is a ending point to these transactions.This commit may be implicit or explicit.  
Example : 
* Begin(Start) 
* Commit  
* Rollback
* Savepoint 
* Set Transaction 


# DCL 
Data Control Language is a subset of SQL that is used to control the access to database. It allow a Database Administrator to grant or revoke permissions on Database Objects, Such as Databases,Tables, Views etc. 
* They interact with databases directly.
* In DCL commands, implicit commit will be there, before and after the command.
* DCL commands deals with priviledges.  
Example of DCL :
1. Grant
2. Revoke
3. Set Role

### Table creation
1. create table student(id int,name varchar(20),age int);
2. desc student; (desc-describe) gives table schema
3. show tables; give table names in current database
4. create table students(id int primary key,name varchar(25),age int); create table with KEY
5. create table employee as select * from student; employee table will have the schema or fields of student. and data also 
6. create table employees1 as select id,name from students; selecting only some fields of a table.
7. create table t1 as select * from students; this will copy data + structure of students.
8. create table t2 as select * from students where true=false; only structure. If a false condition is at the end,it will only copy structure.
* Note : The create table as select (CTS) statement in mysql is used to create a new table by copying data from an existing table or the result of the select query. It creates a new table automatically based on the columns returned by the select statement. It copies selected data,column names and column datatypes and column order.  
The limitation of CTS is that it doesnot copy the table structure rules or database objects.  
The following things are not copied : 
* Primary key
* Foreign key
* Autoincrement property
* Constraints
