## Insert
The insert command in mysql is a dml command which is used to add new recoreds into a table. 
1. Insert values in all columns. 
    ```
    insert into tablename values(value1,value2,value3....)
    ```
    * create table pystudent(id int,name varchar(20),age int,city varchar(50));
    * insert into pystudent values(01,"Harsh",21,"Indore");
    * select * from pystudent;
2. Insert values into specific column.  
If we dont want to insert data into every column, or if we are providing all the values but order is not followed, then we need to provide column list. 
    ```
    insert into tablename(column1,column2,column3) values(value1,value2,value3);
    ```
     insert into pystudent(id,name,city) values(100,"Anil","Dewas"); 
    ``` 
    we have to follow the sequence written along tablename.  
    Note: Best practice always specify the column names while inserting data. This makes the query easier to read. 
3. Insert Multiple Values.
Instead of writing multiple insert statements we can insert many rows in one statement.
    ```
    insert into tablename values(value1,value2,value3...),(value1,value2,value3);
    ```
    * insert into pystudent(id,name,city,age) values(101,"Sahil","Indore",22),(102,"Abhishek","Double choki",22);
4. Insert using Set
MySql also supports the set syntax.
    ```
    insert into tablename set colname=value;
    ```
    * insert into pystudent set id = 105,name = "Bhumi",age=22,city="Indore";
    * insert with set can only one row at a time. 
    * It is not the part of SQL standard, it means db like oracle,Sql server and PostgreSql do not support this syntax.
5. Insert data from Another Table
We can copy records from one table to another table.
    ```
    create table (new name) as select * from oldtable where true=false;
    insert into (new) select * from (old);
    ```
    * 1. create table pybackup as select * from pystudent where 1=0; ( this creates a new table with same structure)
      2. insert into pybackup select * from pystudent;( this copies all data)
      3. insert into pybackup(id,name) select id,name from pystudent; (copies onlu id , name and insert null into all others).  
       
    Note :
    1. Values must match with column order.
    2. Wrong order can lead to error.
    3. No of values must match with no of cols.

    
    





## Primary Key 
A primary key is a column or group of columns, whose values uniquely identify each row in a table. It doesn't allow duplicates or null values.  
* without a primary key duplicate records can exist and it become difficult to identify a particular row.
* Relationships between tables cannot be established properly.
* Data Integrity will be reduced.
* Searching records become less reliable.
### Characteristics: 
1. Unique values
2. Cannot contain NULL
3. Only one primary key per table is allowed.However that primary key can consist multiple columns(composite primary key)
4. It automatically creates an Index:when a Primary key is created MySql automatically creates a unique index on that column.
5. Values of a primary key should be stable.
### Ways
```
1. * create table pyemployee(empid int primary key,name varchar(22),salary decimal(10,2));
   * desc pyemployee
2.  create table pyemployee1(empid int,name varchar(22),salary decimal(10,2),constraint pk_employee primary key(empid)); (pk_employee is a constraint name).
3. After table creation if we need to add primary key
   * create table pyemployee3(empid int,name varchar(22),salary decimal(10,2));
   * alter table pyemployee3 add primary key(empid);
   1. insert into pyemployee values(101,"Deepika",500000),(102,"rashmika",700000);
   2. insert into pyemployee(name,salary) values('thapaji',5000); This will raise error as id is primary key.
```

### Autoincrement Property 
It is a MySql attribute used with numeric columns to automatically generate a unique no whenever a new record is inserted into a table. It is most commonly used with the primary key so that the user doesn't have to enter the id manually.  
Whenever a new employee joins we must manually decide the next id.  
Problems:
* We may forget the id 
* Duplicate id's can be inserted accidentally.

To solve these autoincrement is used. 
``` 
    create table pyemployee5(id int auto_increment primary key,name varchar(20),salary decimal(10,2));
    desc pyemployee5;
    insert into pyemployee5(name,salary) values("anil",5000);
```  
Rules for Autoincrement : 
1. Only one autoincrement column in the table.

**Default Starting Value** : By default autoincrement starts from 1.and by default it increases by 1.  
Changing the starting value : we can change the starting value of autoincrement. 
* create table pyemployee6(eid int auto_increment primary key,name varchar(20)) auto_increment=1000;
* insert into pyemployee6(name) values("ani");

**Manually Inserting Autoincrement values**: 
  * create table pyemployee7(eid int auto_increment primary key,name varchar(50));
  * insert into pyemployee7(name) values("katappa")
  * insert into pyemployee7 values(10,"Deepika")
  * insert into pyemployee7(name) values("rashmika")  

  The value if we manually increment it will then auto_increment from that manual value. If we manually insert the value that is greater than the current autoincrement value  then MySql will update the next autoincrement automatically.  
  *  insert into pyemployee7 values(2,"Anil");
  * insert into pyemployee7(name) values("Sahil"); 

  Like this 2 value Anil entry will be done but when we again use autoincrement type query then it will start from 12 Sahil.   

**Autoincrement with NULL Value** : If we explicitly insert NULL value into an autoincrement column then MySql treats it as we didn't provided any value, and auto generates the next sequence number.
* insert into pyemployee7 values(null,"abc"); It gives 13 no to abc. 

**Autoincrement with 0** : Same it also increases count , If we insert 0 into an autoincrement column then the result depends on MySql SQL Mode.
1. Case 1: Default behaviour  
    In most MySql installations if the Sql Mode doesn't include N0_AUTO_VALUE_ON_ZERO then inserting zero into an autoincrement column behave exactly like inserting null.It means mysql auto generates the next available value. 
    *  insert into pyemployee7 values(0,"Abhi"); It gives 14 no to abhi.
2. Case 2: When N0_AUTO_VALUE_ON_ZERO mode is enabled then 0 is treated as a normal value. 
  **Use select @@sql mode;** to check the Mode of MySql. 

**Auto_increment gap** : There are several reasons why gaps can occur.
1. Case 1: If you perform delete after inserting some value like (1,2,3) and you delete 2 entry. Then the deleted entry will never be reused in auto_increment.

### Why truncate is called as DDL command 
Truncate is considered  as DDL because it is treated as a table-level structural operation. Not as a row-by-row dml operation. It operates at the table level and deallocates/resets the table's data storage rather than processing individual row like delete.  
* truncate table student where student_id=2; **INVALID**  
1. Truncate doesn't work with where clause;
2. It doesn't delete table structure
3. It resets Autoincrement values , but in case of delete auto_increment doesn't get reset.
    ```
    create table with auto increment
    insert data 
    then use truncate 
    then insert again   
    the auto increment starts from 1 again.
    ```
**When should we use Truncate** : * When we don't need any existing rows. 
                                  * When table is temporary table.
                                  * Truncate is faster than delete.

**Truncate and Transaction** : Truncate table cause an explicit commit and cannot be rolled back. 

**Truncate and Triggers** : truncate command doesn't fire delete trigger.

## Drop command
Drop is a DDL command which is used to permanently remove a database object from the Db.  
When we use Drop the object gets deleted along with the data stored inside it. 
```
Drop table tablename;
```
ex : drop table pyemployee1;
* After Drop command the table will no longer exist. 
### Drop multiple tables 
``` 
drop table tablen1,tablen2,tablen3;
```
*  drop table pyemployee;
* Again if we do it give error ( drop table pyemployee;)
so we use 
* drop table if exists pyemployee;
 

## Truncate VS Drop
1. * Truncate is used to remove all rows while keeping the table. 
   * Drop is used to remove entire table.
2. * In truncate data removed but structure retained.
   * In Drop both gets removed.
3. * Truncate resets the auto_increment counter. 
   * Drop the table itself is removed including its auto_increment.
4. * Truncate doesn't support where clause.
   * Drop also doesn't support where bcoz it remove entire table. 
5. Both are ddl operations so they can't be rolled back.


## Alter command
Alter is a DDL command which is used to modify the structure of an existing db object. It allows us to change an existing table without droping and recreating the entire table. 

We use alter for:
* add a column 
* drop a column 
* modify column's datatype
* change column definition 
* rename a column 
* add constraints
* Drop constraints
* Add/Drop primary key
* Add/Drop foreign key
* change auto_increment etc. 

```
alter table tablename alteroperation;
```  
**Note**: Alter table is one of the mechanism used to evolve the database schema.

* create table Infobeans(st_id int,name varchar(20));
