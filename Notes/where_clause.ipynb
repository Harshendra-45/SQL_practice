select * from employee_batch where city="Indore" or city="Pune" and salary>50000;
+------+-------+------+--------+------------+----------+
| id   | name  | age  | city   | department | salary   |
+------+-------+------+--------+------------+----------+
|    1 | Rahul |   25 | Indore | IT         | 45000.20 |
+------+-------+------+--------+------------+----------+
1 row in set (0.00 sec)
```
Note: In MySql Precedences are:
1. Not
2. And
3. or  
It means whenever we write **where a or b and c** then it will be interpreted as **where a or (b and c)** .
1. write a query to fetch employees who are from Indore or Bhopal, And their salaries must be greater than 40k;
```

By default MySql Sorts coln in Ascending order(Asc).
* OrderBy doesn't change the actual data stored in the table, It just changes the order in which the result is being displayed. 
    ``` 
    select * from employee_batch order by salary;
    select * from employee_batch order by salary desc ;
    select department from employee_batch order by salary desc; We can get specific points 
    select * from employee_batch order by city desc; if null is there in last if desc and in first if asc. 
    ``` 
1. WAQ to select all the it employee and display them from high to low salary.  
    select * from employee_batch where department="it" order by salary desc;
Note: In the Above Query where decides which rows we want and orderby decides in which order we want those rows. 

### Order by with multiple columns
*  select * from employee_batch order by city,salary desc;
It will order city in ascending then go to second priority if two duplicates exist and then decide the result by salary in descending order.1st priority = city, 2nd priority = salary
* select * from employee_batch order by city desc,salary desc;
here both will follow descending order to present result.
